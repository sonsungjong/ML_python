```python
import pandas as pd

df = pd.read_csv('mdi.csv')
df.head()
```

```python
df.HUMAN.value_counts()
```

```python
# NULL이 없는지 확인
df.HUMAN.info()
```

```python
# 1단계: 0 또는 1이면 0으로 통일
df.loc[df['HUMAN'] <= 1, 'HUMAN'] = 0
df['HUMAN'].value_counts()
```

```python
# 2단계: 2 이상인 것들은 모두 2로 만들기
df.loc[df['HUMAN'] >= 2, 'HUMAN'] = 2
df['HUMAN'].value_counts()
```

```python
# 3단계: 2를 모조리 1로 변경
df.loc[df['HUMAN'] == 2, 'HUMAN'] = 1
df['HUMAN'].value_counts()
```

```python
# 최종 확인 및 저장
print(df['HUMAN'].unique())  # [0, 1]만 있어야 함
df.to_csv('mdi.csv', index=False)
print('저장 완료')
```
