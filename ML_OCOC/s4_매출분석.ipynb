```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from statsmodels.stats.outliers_influence import variance_inflation_factor

np.random.seed(42)
```

```python
# 온라인 쇼핑몰 종합 분석 - 매출 증대를 위한 데이터 기반 의사결정

# 데이터 준비
n_customers = 2000

# 타겟 특징
age = np.random.normal(35, 12, n_customers)
age = np.clip(age, 18, 70)
gender = np.random.choice(['M', 'F'], n_customers, p=[0.45, 0.55])
region = np.random.choice(['North', 'South', 'East', 'West'], n_customers)
member_tier = np.random.choice(['Basic', 'Silver', 'Gold'], n_customers, p=[0.6, 0.3, 0.1])

# 행동 데이터
visit_frequency = np.random.poisson(3, n_customers)     # 월 방문 횟수
page_views = visit_frequency * np.random.uniform(5, 15, n_customers)
cart_abandonment = np.random.binomial(1, 0.3, n_customers)      # 장바구니 이탈

# 마케팅 데이터
email_subscriber = np.random.binomial(1, 0.4, n_customers)
promo_code_used = np.random.binomial(1, 0.25, n_customers)
marketing_channel = np.random.choice(['Organic', 'Social', 'Email', 'Paid'], n_customers, p=[0.3, 0.25, 0.2, 0.25])

# 구매 데이터
# 구매 확률 모델링
purchase_prob = (
    0.2 
    + 0.01 * visit_frequency 
    + 0.1 * email_subscriber 
    + 0.15 * promo_code_used 
    + 0.1 * (member_tier == 'Gold') 
    + 0.05 * (member_tier == 'Silver') 
    - 0.1 * cart_abandonment
)
purchase_prob = np.clip(purchase_prob, 0, 1)
made_purchase = np.random.binomial(1, purchase_prob)

# 구매 금액 (구매한 사람만)
purchase_amount = np.where(made_purchase == 1,
    np.random.gamma(2, 50, n_customers) *
    (1 + 0.2 * (member_tier == 'Gold') + 0.1 * (member_tier == 'Silver')),
    0)

# 데이터프레임 생성
ecomerce_data = pd.DataFrame({
    'customer_id': range(1, n_customers + 1),
    'age': age,
    'gender': gender,
    'region': region,
    'member_tier': member_tier,
    'visit_frequency': visit_frequency,
    'page_views': page_views,
    'cart_abandonment': cart_abandonment,
    'email_subscriber': email_subscriber,
    'promo_code_used': promo_code_used,
    'marketing_channel': marketing_channel,
    'made_purchase': made_purchase,
    'purchase_amount': purchase_amount
})

print("=" * 80)
print('온라인 쇼핑몰 데이터 분석 프로젝트')
print("=" * 80)
print(f"\n총 고객 수: {n_customers} 명")
```
