```python
# 결측치만 조회
df[df.isna().any(axis=1)]
```

```python
# dropna() : nan 있으면 행 제거

# 학생과 클래스가 없을떄만 제거할땐
df[df.isna().any(axis=1)].dropna(subset=['Student', 'Class'])
```

```python
df[df.Grade.isna()]
```

```python
df.Grade.mean()
```

```python
# inplace : 원본에 반영
df.Grade.fillna(df.Grade.mean(), inplace=True)
```

```python
df.Grade
```

```python
df[df.isna().any(axis=1)]
```

```python
df[df.Class=='Freshman Seminar']
```

```python
# Freshman Seminar 클래스는 신입생일 것이기 때문에 Freshman 으로 채우자
df.Year = np.where(df.Year.isna(), 'Freshman', df.Year)
df
```
