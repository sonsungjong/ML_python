```python
# 문자열이면 계산이나 평균이 안나오기 때문에 숫자나 날짜로 변환한다
# 방해되는 $ 나 , 같은 문자는 replace 로 제거한다
run_times.Fee = pd.to_numeric(run_times.Fee.str.replace('$', '').str.replace(',', ''))
```

```python
run_times['Warm Up Time'] = pd.to_numeric(run_times['Warm Up Time'].astype('str').str.replace(' min', ''))
```
