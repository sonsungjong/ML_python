```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./Data Files/Movie_classification.csv', header = 0)
```

```python
df.head()
```

```python
df.info()
```

```python
# 결측치 전처리
time_avg = df['Time_taken'].mean()
df['Time_taken'].fillna(value=time_avg, inplace=True)
```

```python
df.info()
```

```python
# 더미값 생성 (문자열 변수를 머신러닝이 처리할 수 있는 숫자형 특성으로 변환한다)
df = pd.get_dummies(df, columns=['3D_available', 'Genre'], drop_first=True)
```

```python
df.head()
```

```python
# X 와 y 로 나눈다 (특성과 결과)
X = df.loc[:, df.columns != "Start_Tech_Oscar"]
X.head()
```

```python
X.shape
```

```python
y = df["Start_Tech_Oscar"]
y.head()
```

```python
y.shape
```

```python
# X 와 y를 생성했으면 학습용과 테스트용으로 Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0) # 실제는 거의 42
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
# 분류 모델 훈련시키기 (과적합 / 과소적합 여부 확인)
from sklearn import tree
clf_tree = tree.DecisionTreeClassifier(max_depth=3)
clf_tree.fit(X_train, y_train)
```

```python
# 정확도 테스트를 위해 예측(predict) 해보기
y_train_prediction = clf_tree.predict(X_train)
y_test_prediction = clf_tree.predict(X_test)
```

```python
y_test_prediction
```

```python
# 모델 Performance
from sklearn.metrics import accuracy_score, confusion_matrix

confusion_matrix(y_train, y_train_prediction)       # 실제와 예측
```

```python
confusion_matrix(y_test, y_test_prediction)
```

```python
accuracy_score(y_test, y_test_prediction)  # 정확도
```

```python
# 시각화 (Decision Tree)
from IPython.display import Image
import pydotplus

dot_data = tree.export_graphviz(clf_tree, out_file=None, feature_names= X_train.columns, filled=True)
```

```python
graph = pydotplus.graph_from_dot_data(dot_data)
Image(graph.create_png())
```
